Q1. Sum of First 10 Natural Numbers

```python
# Q1: Find Sum of First 10 Natural Numbers

sum_numbers = 0

for i in range(1, 11):
    sum_numbers += i

print("Sum of first 10 natural numbers =", sum_numbers)
```

Q2. Factorial of a Number

```python
# Q2: Find Factorial of a Number

num = int(input("Enter a number: "))

factorial = 1

for i in range(1, num + 1):
    factorial *= i

print("Factorial of", num, "=", factorial)
```

Q3. Fibonacci Series

```python
# Q3: Print Fibonacci Series

n = int(input("Enter the number of terms: "))

a = 0
b = 1

print("Fibonacci Series:")

for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

Q4. Find Largest Among 3 Numbers

```python
# Q4: Find Largest Among 3 Numbers

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
num3 = float(input("Enter third number: "))

if num1 >= num2 and num1 >= num3:
    largest = num1
elif num2 >= num1 and num2 >= num3:
    largest = num2
else:
    largest = num3

print("Largest number =", largest)
```

Q5. Student Result System
User input
if-elif-else
Loops
Percentage calculation
Grade display

```python
# Q5: Student Result System

print("----- Student Result System -----")

name = input("Enter Student Name: ")
roll_no = input("Enter Roll Number: ")

total_marks = 0
subjects = 5

for i in range(1, subjects + 1):
    marks = float(input(f"Enter marks for Subject {i}: "))
    total_marks += marks

percentage = total_marks / subjects

if percentage >= 90:
    grade = "A+"
elif percentage >= 80:
    grade = "A"
elif percentage >= 70:
    grade = "B"
elif percentage >= 60:
    grade = "C"
elif percentage >= 50:
    grade = "D"
else:
    grade = "F"

print("\n----- Result -----")
print("Student Name :", name)
print("Roll Number  :", roll_no)
print("Total Marks  :", total_marks)
print("Percentage   :", round(percentage, 2), "%")
print("Grade        :", grade)
```
