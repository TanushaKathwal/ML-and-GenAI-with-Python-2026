Q1. Area of Rectangle

```python
# Q1: Find Area of Rectangle

length = float(input("Enter length: "))
breadth = float(input("Enter breadth: "))

area = length * breadth

print("Area of Rectangle =", round(area,2), "sq. units")
```

Q2. Simple Interest

```python
# Q2: Find Simple Interest

p = float(input("Enter Principal: "))
r = float(input("Enter Rate: "))
t = float(input("Enter Time: "))

si = (p * r * t) / 100

print("Simple Interest =", si)
```

Q3. Celsius to Fahrenheit

```python
# Q3: Celsius to Fahrenheit

celsius = float(input("Enter temperature in Celsius: "))

fahrenheit = (celsius * 9/5) + 32

print("Temperature in Fahrenheit =", fahrenheit)
```

Q4. Average of 3 Numbers

```python
# Q4: Average of 3 Numbers

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
c = float(input("Enter third number: "))

average = (a + b + c) / 3

print("Average =", round(average))
```

Q5. Square and Cube

```python
# Q5: Square and Cube

num = int(input("Enter a number: "))

print("Square =", num ** 2)
print("Cube =", num ** 3)
```

Q6. Swap Without Third Variable

```python
# Q6: Swap Two Numbers

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

a, b = b, a

print("After Swapping:")
print("a =", a)
print("b =", b)
```

Q7. Student Report Program

```python
# Q7: Student Report Program

name = input("Enter Student Name: ")
roll_no = input("Enter Roll Number: ")

mark1 = float(input("Enter Marks in Subject 1: "))
mark2 = float(input("Enter Marks in Subject 2: "))
mark3 = float(input("Enter Marks in Subject 3: "))

total = mark1 + mark2 + mark3
percentage = total / 3

print("\n----- Student Report -----")
print("Name:", name)
print("Roll No:", roll_no)
print("Total Marks:", total)
print("Percentage:", round(percentage), "%")
```
